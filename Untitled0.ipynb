```python
import pandas as pd

matches = pd.read_csv("matches.csv")
deliveries = pd.read_csv("deliveries.csv")

print(matches.head())
print(deliveries.head())
```

```python
matches.isnull().sum()

matches = matches.dropna()
```

```python
team_wins = matches['winner'].value_counts()

print(team_wins)
```

```python
team_wins.plot(kind='bar')
```

```python
toss_win_match_win = matches[
    matches['toss_winner'] == matches['winner']
]

percentage = (
    len(toss_win_match_win) / len(matches)
) * 100

print(percentage)
```

```python
venue_wins = matches.groupby('venue')['winner'].count()

print(venue_wins.sort_values(ascending=False))
```

```python
top_batsman = deliveries.groupby('batter')['batsman_runs'].sum()

print(top_batsman.sort_values(ascending=False).head(10))
```

```python
top_bowlers = deliveries['bowler'].value_counts()

print(top_bowlers.head(10))
```

```python
matches.to_csv("clean_matches.csv", index=False)
deliveries.to_csv("clean_deliveries.csv", index=False)
```

```python
import pandas as pd
import sqlite3
```

```python
matches = pd.read_csv("matches.csv")
deliveries = pd.read_csv("deliveries.csv")

print(matches.head())
print(deliveries.head())
```

```python
conn = sqlite3.connect("ipl.db")
```

```python
matches.to_sql(
    "matches",
    conn,
    if_exists="replace",
    index=False
)

deliveries.to_sql(
    "deliveries",
    conn,
    if_exists="replace",
    index=False
)

print("Tables created successfully")
```

```python
query = """
SELECT winner, COUNT(*) as wins
FROM matches
GROUP BY winner
ORDER BY wins DESC
"""

result = pd.read_sql(query, conn)

print(result)
```

```python
query = """
SELECT COUNT(*) as toss_and_match_win
FROM matches
WHERE toss_winner = winner
"""

result = pd.read_sql(query, conn)

print(result)
```

```python
query = """
SELECT venue, COUNT(*) as total_matches
FROM matches
GROUP BY venue
ORDER BY total_matches DESC
"""

result = pd.read_sql(query, conn)

print(result)
```

```python
result.to_csv("team_wins.csv", index=False)
```

```python
query = """
SELECT batter,
SUM(batsman_runs) as total_runs
FROM deliveries
GROUP BY batter
ORDER BY total_runs DESC
LIMIT 10
"""

result = pd.read_sql(query, conn)

print(result)
```
